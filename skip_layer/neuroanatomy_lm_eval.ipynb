# Neuroanatomy Skip Layer
    

## Visualization Code

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

def plot(model_name, layer_num, task_category, tasks):
    result_dir = Path(f"eval-results/{model_name}/{task_category}")
    base_df = pd.read_csv(result_dir / "base.csv")

    all_results = []

    for layer_idx in range(layer_num):

        csv_path = result_dir / f"skip_layer_{layer_idx}.csv"

        if not csv_path.exists():
            print(f"Missing: {csv_path}")
            continue

        df = pd.read_csv(csv_path)

        df["layer"] = layer_idx

        all_results.append(df)

    skip_df = pd.concat(all_results, ignore_index=True)

    for task_name in tasks:

        base_task = base_df[base_df["task"] == task_name]

        if len(base_task) == 0:
            print(f"Task not found: {task_name}")
            continue

        base_acc = float(base_task.iloc[0]["acc"])

        task_df = skip_df[skip_df["task"] == task_name]

        if len(task_df) == 0:
            print(f"No results: {task_name}")
            continue

        task_df = task_df.sort_values("layer")

        layers = task_df["layer"]
        accs = task_df["acc"]

        plt.figure(figsize=(14, 5))

        plt.bar(layers, accs)

        plt.axhline(
            y=base_acc,
            linestyle="--",
            label=f"Base",
        )

        short_name = task_name.replace(
            "global_mmlu_full_id_", ""
        )

        plt.title(f"{task_category}_{short_name} on {model_name}")

        plt.xlabel("Skipped Layer")
        plt.ylabel("Accuracy")

        plt.xticks(range(layer_num))

        plt.legend()

        plt.show()
```
